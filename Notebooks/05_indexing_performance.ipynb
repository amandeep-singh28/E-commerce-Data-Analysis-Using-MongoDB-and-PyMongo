## Analysis 4: Indexing and Query Performance

### Objective

The objective of this analysis is to examine how indexing affects query execution in MongoDB. An index will be created on the product price field, and query execution statistics will be compared before and after indexing.

```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
MONGODB_URI = os.getenv("MONGODB_URI")

if not MONGODB_URI:
    raise ValueError("MONGODB_URI is missing from the .env file")
```

```python
client = MongoClient(MONGODB_URI)
```

```python
db = client["ecommerce_analysis"]
products_collection = db["products"]

print("Successfully connected to MongoDB Atlas!")
print("Database:", db.name)
print("Collection:", products_collection.name)
```

### Measure Query Performance Before Indexing

Run an `explain` command with execution statistics for products priced above 100. Record returned and examined document counts and execution time as the baseline.

```python
query = {"price": {"$gt": 100}}

before_index = db.command(
    "explain",
    {
        "find": "products",
        "filter": query
    },
    verbosity="executionStats"
)

before_stats = before_index["executionStats"]

print("BEFORE INDEXING")
print("-" * 30)
print("Documents returned:", before_stats["nReturned"])
print("Documents examined:", before_stats["totalDocsExamined"])
print("Execution time (milliseconds):", before_stats["executionTimeMillis"])
```

### Create an Index on Price

Create an ascending index on the `price` field so MongoDB can use it to support price-based queries.

```python
index_name = products_collection.create_index([("price", 1)])

print("Index created successfully!")
print("Index name:", index_name)
```

### Measure Query Performance After Indexing

Run the same query again with execution statistics and compare its document counts and execution time with the baseline.

```python
after_index = db.command(
    "explain",
    {
        "find": "products",
        "filter": query
    },
    verbosity="executionStats"
)

after_stats = after_index["executionStats"]

print("AFTER INDEXING")
print("-" * 30)
print("Documents returned:", after_stats["nReturned"])
print("Documents examined:", after_stats["totalDocsExamined"])
print("Execution time (milliseconds):", after_stats["executionTimeMillis"])
```
