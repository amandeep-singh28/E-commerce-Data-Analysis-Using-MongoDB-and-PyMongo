### Import PyMongo and Utilities

```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os
from getpass import getpass
```

```python
load_dotenv()
```

### Connect to MongoDB

```python
MONGODB_URI = os.getenv("MONGODB_URI")

if not MONGODB_URI:
    raise ValueError("MONGODB_URI is missing from the .env file")
```

```python
client = MongoClient(MONGODB_URI)
```

### Analysis 1: Product Data Exploration and Filtering

### 1. Objective

The objective of this analysis is to explore product records stored in MongoDB and demonstrate document retrieval, projection, filtering, sorting, and limiting operations using PyMongo.

### Select the Database and Collection

```python
db = client["ecommerce_analysis"]

products_collection = db["products"]

print("Database:", db.name)
print("Collection:", products_collection.name)
```

### Count Product Records

```python
total_products = products_collection.count_documents({})

print("Total number of products:", total_products)
```

### Preview Product Documents

```python
for product in products_collection.find({}, {"_id" : 0}).limit(5):
    print(product)
```

### Project Selected Product Fields

```python
for product in products_collection.find(
    {},
    {
        "_id": 0,
        "id": 1,
        "name": 1,
        "price": 1,
        "is_active": 1
    }
).limit(10):
    print(product)
```

### Filter Products by Price

```python
expensive_products = products_collection.find(
    {"price": {"$gt": 100}},
    {
        "_id": 0,
        "id": 1,
        "name": 1,
        "price": 1
    }
)

for product in expensive_products.limit(10):
    print(product)
```

### Retrieve Active Products

```python
active_products = products_collection.find(
    {"is_active": True},
    {
        "_id": 0,
        "id": 1,
        "name": 1,
        "price": 1,
        "is_active": 1
    }
)

for product in active_products.limit(10):
    print(product)
```

### Compare Active and Inactive Products

```python
active_count = products_collection.count_documents(
    {"is_active": True}
)

inactive_count = products_collection.count_documents(
    {"is_active": False}
)

print("Active products:", active_count)
print("Inactive products:", inactive_count)
```

### Find the Cheapest Products

```python
from pymongo import ASCENDING, DESCENDING

cheapest_products = products_collection.find(
    {},
    {
        "_id": 0,
        "id": 1,
        "name": 1,
        "price": 1
    }
).sort("price", ASCENDING).limit(5)

for product in cheapest_products:
    print(product)
```

### Find the Most Expensive Products

```python
most_expensive_products = products_collection.find(
    {},
    {
        "_id": 0,
        "id": 1,
        "name": 1,
        "price": 1
    }
).sort("price", DESCENDING).limit(5)

for product in most_expensive_products:
    print(product)
```

### Summarize Product Counts

```python
total_products = products_collection.count_documents({})
active_count = products_collection.count_documents({"is_active": True})
inactive_count = products_collection.count_documents({"is_active": False})

print("PRODUCT ANALYSIS SUMMARY")
print("-" * 35)
print("Total products:", total_products)
print("Active products:", active_count)
print("Inactive products:", inactive_count)
```
