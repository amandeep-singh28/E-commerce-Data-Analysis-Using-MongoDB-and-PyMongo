## E-commerce Data Analysis Using MongoDB and PyMongo

### 1. Database Connection

This section establishes a connection between Python and MongoDB Atlas using PyMongo.

```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os
from getpass import getpass
```

```python
load_dotenv()
```

```python
MONGODB_URI = os.getenv("MONGODB_URI")

if not MONGODB_URI:
    raise ValueError("MONGODB_URI is missing from the .env file")
```

```python
client = MongoClient(MONGODB_URI)
```

### Select the Database

```python
db = client["ecommerce_analysis"]

print("Database selected:", db.name)
```

### Locate the Dataset File

```python
from pathlib import Path

DATA_PATH = (
    Path.cwd().parent
    / "Data"
    / "ecommerce-sample-database-1k-en_GB-v2.json"
)

print("Dataset path:", DATA_PATH)
print("File exists:", DATA_PATH.exists())
```

### Load and Inspect the JSON Data

```python
import json

with open(DATA_PATH, "r", encoding="utf-8") as file:
    data = json.load(file)

print("Data type:", type(data).__name__)

if isinstance(data, dict):
    print("Top-level keys:")
    for key in data.keys():
        print("-", key)

elif isinstance(data, list):
    print("Number of top-level records:", len(data))
```

### Import Data into MongoDB

```python
from pymongo.errors import PyMongoError

import_results = {}

for collection_name, records in data.items():

    if not isinstance(records, list):
        print(f"Skipping {collection_name}: data is not a list")
        continue

    if not records:
        print(f"Skipping {collection_name}: no records")
        continue

    collection = db[collection_name]

    result = collection.insert_many(records)

    import_results[collection_name] = len(result.inserted_ids)

    print(f"{collection_name}: {len(result.inserted_ids)} documents inserted")

print("\nImport completed!")
```

### Verify the Imported Collections

```python
collection_names = db.list_collection_names()

print("Total collections imported:", len(collection_names))

for collection_name in sorted(collection_names):
    print(
        f"{collection_name}: "
        f"{db[collection_name].count_documents({})} documents"
    )
```
