```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
MONGODB_URI = os.getenv("MONGODB_URI")

if not MONGODB_URI:
    raise ValueError("MONGODB_URI is missing from the .env file")
```

```python
client = MongoClient(MONGODB_URI)
```

```python
db = client["ecommerce_analysis"]
```

```python
products = db["products"]
brands = db["brands"]
categories = db["categories"]
```

```python
for i in products.find(
    {

    },
    {
        "_id" : 0
    }
).limit(2):
    print(i)
```

```python
for i in brands.find(
    {

    },
    {
        "_id" : 0,
        "logo_url" : 0
    }
).limit(2):
    print(i)
```

```python
for i in categories.find(
    {

    },
    {
        "_id" : 0
    }
).limit(2):
    print(i)
```

```python
orders = db["orders"]
addresses = db["addresses"]
```

```python
orders.count_documents({})
```

```python
addresses.count_documents({})
```

```python
for i in orders.find(
    {

    },
    {
        "_id" : 0
    }
):
    print(i)
```

```python
for i in orders.find(
    {

    },
    {
        "_id" : 0,
        "billing_address_id" : 1
    }
).limit(2):
    print(i)
```

```python
pipeline = [
    {
        "$group": {
            "_id": None,
            "min_id": {"$min": "$billing_address_id"},
            "max_id": {"$max": "$billing_address_id"}
        }
    }
]

result = list(orders.aggregate(pipeline))
if result:
    print(f"Min: {result[0]['min_id']}, Max: {result[0]['max_id']}")
```
